```python
available_chocolate=5 # the machine has only 5 chocolates

x=int(input('How many chocolates you want?'))

i=1
while i<=x:
    if i>available_chocolate: # we stop the execution but which code execution not entire code , i want to come out of while loop 
        break # break is statement | means jump out of the loop
    print('Chocolate')
    i+=1

print('Bye for now')
```
